```python
import easyocr

from transformers import pipeline

from PIL import Image

import matplotlib.pyplot as plt
```

```python
reader = easyocr.Reader(['en'])

print("OCR Model Loaded Successfully")
```

```python
sentiment_pipeline = pipeline(
    "sentiment-analysis"
)

print("Sentiment Model Loaded")
```

```python
import os

# Folder path
folder_path = "test_images"

all_reviews = []

# Read every image
for image_file in os.listdir(folder_path):

    image_path = os.path.join(
        folder_path,
        image_file
    )

    # OCR
    result = reader.readtext(
        image_path,
        detail=0
    )

    # Join text
    text = " ".join(result)

    # Store reviews
    all_reviews.append(text)

# Combine all reviews
combined_reviews = " ".join(all_reviews)

print(combined_reviews)
```

```python
sentiment = sentiment_pipeline(
    combined_reviews[:512]
)

print(sentiment)
```

```python
import nltk

nltk.download('punkt')

nltk.download('punkt_tab')

from nltk.tokenize import sent_tokenize
```

```python
sentences = sent_tokenize(
    combined_reviews
)

print(sentences[:5])
```

```python
positive_sentences = []

negative_sentences = []

for sentence in sentences:

    result = sentiment_pipeline(
        sentence[:512]
    )

    label = result[0]['label']

    if label == "POSITIVE":

        positive_sentences.append(
            sentence
        )

    else:

        negative_sentences.append(
            sentence
        )
```

```python
print("=" * 50)

print("TOP POSITIVE INSIGHTS 😎\n")

for sentence in positive_sentences[:5]:

    print("-", sentence)
```

```python
print("=" * 50)

print("TOP NEGATIVE INSIGHTS 😭\n")

for sentence in negative_sentences[:5]:

    print("-", sentence)
```

```python
print("=" * 60)

print("FINAL PRODUCT REVIEW ANALYSIS 😎")

print("\nOverall Sentiment:")
print(sentiment[0]['label'])

print("\nConfidence Score:")
print(
    round(
        sentiment[0]['score'] * 100,
        2
    ),
    "%"
)

print("\nTop Positive Insights ✅")

for sentence in positive_sentences[:5]:

    print("-", sentence)

print("\nTop Negative Insights ❌")

for sentence in negative_sentences[:5]:

    print("-", sentence)

# ====================================
# FINAL RECOMMENDATION
# ====================================

if len(positive_sentences) > len(negative_sentences):

    recommendation = "Worth Buying ✅"

else:

    recommendation = "Not Recommended ❌"

print("\nFinal Recommendation:")
print(recommendation)

print("=" * 60)
```

```python
# ====================================
# PRODUCT RATING SYSTEM
# ====================================

positive_count = len(
    positive_sentences
)

negative_count = len(
    negative_sentences
)

total = (
    positive_count +
    negative_count
)

# Avoid division by zero
if total == 0:

    rating = 3.0

else:

    positivity_ratio = (
        positive_count / total
    )

    rating = round(
        positivity_ratio * 5,
        1
    )

print("=" * 50)

print("AI PRODUCT RATING ⭐\n")

print(f"Overall Rating: {rating}/5 ⭐")

print("=" * 50)
```
